### Setup DistilBERT for Ticket Classification

This section will set up the DistilBERT model to classify incoming customer support tickets. This classification will then be used by the routing engine to direct tickets to the appropriate department.

```python
# Install necessary libraries
!pip install transformers torch
```

Next, we'll define a function to load the pre-trained DistilBERT model and its tokenizer. For demonstration purposes, we will use a general sequence classification model, but in a real-world scenario, this would be a fine-tuned model specific to your ticket categories.

```python
from transformers import DistilBertTokenizer, DistilBertForSequenceClassification
import torch

# Define model name (you would use your fine-tuned model here)
MODEL_NAME = 'distilbert-base-uncased'

def load_distilbert_model_and_tokenizer(model_name=MODEL_NAME):
    """Loads a pre-trained DistilBERT model and tokenizer."""
    tokenizer = DistilBertTokenizer.from_pretrained(model_name)
    # For a real routing system, this would be a fine-tuned model
    # For demonstration, we'll use a generic sequence classification model.
    # In a real scenario, you'd load your fine-tuned model with specific labels.
    model = DistilBertForSequenceClassification.from_pretrained(model_name, num_labels=3) # Example: 3 labels for categories
    return tokenizer, model

print("DistilBERT model and tokenizer loading function defined.")
```

Now, let's create a function to classify the redacted ticket text. This function will take the text, tokenize it, and use the DistilBERT model to predict a category. It will also simulate sentiment detection, as mentioned in your project document.

```python
def classify_ticket(text: str, tokenizer, model):
    """Classifies a ticket's category and sentiment using DistilBERT."""
    # Tokenize the input text
    inputs = tokenizer(text, return_tensors='pt', truncation=True, padding=True)

    # Get model outputs
    with torch.no_grad():
        outputs = model(**inputs)

    # Get predicted class (category)
    logits = outputs.logits
    predicted_class_id = torch.argmax(logits, dim=1).item()

    # Map class ID to a meaningful category label (this would be based on your fine-tuning)
    # For this example, let's assume a simple mapping:
    category_labels = {
        0: "Technical Support",
        1: "Billing Inquiry",
        2: "General Feedback"
    }
    predicted_category = category_labels.get(predicted_class_id, "Uncategorized")

    # Simulate sentiment detection (DistilBERTForSequenceClassification can be fine-tuned for this too)
    # For simplicity, we'll assign a random sentiment or base it on a simple rule.
    if "urgent" in text.lower() or "issue" in text.lower():
        sentiment = "negative"
    elif "thank you" in text.lower() or "good" in text.lower():
        sentiment = "positive"
    else:
        sentiment = "neutral"

    # Confidence score (example: using softmax probability of the predicted class)
    probabilities = torch.softmax(logits, dim=1)
    confidence = probabilities[0][predicted_class_id].item()

    return {"category": predicted_category, "sentiment": sentiment, "confidence": round(confidence, 3)}

print("Ticket classification function defined.")
```

Finally, let's demonstrate how to use these functions to classify a sample ticket.

```python
# Load the tokenizer and model
tokenizer, model = load_distilbert_model_and_tokenizer()

# Sample redacted ticket text
sample_ticket_text_1 = "My internet is not working. I have tried restarting the router multiple times. This is urgent!"
sample_ticket_text_2 = "I have a question about my last month's bill. It seems incorrect."
sample_ticket_text_3 = "Just wanted to give some general feedback on your service. It was great!"

# Classify the sample tickets
classification_1 = classify_ticket(sample_ticket_text_1, tokenizer, model)
classification_2 = classify_ticket(sample_ticket_text_2, tokenizer, model)
classification_3 = classify_ticket(sample_ticket_text_3, tokenizer, model)

print(f"Ticket 1: '{sample_ticket_text_1}'\n  Classification: {classification_1}\n")
print(f"Ticket 2: '{sample_ticket_text_2}'\n  Classification: {classification_2}\n")
print(f"Ticket 3: '{sample_ticket_text_3}'\n  Classification: {classification_3}\n")
```

### Implementing SLA Deadline Assignment

As per your project's architecture (Section 8.5), the next step is to assign an SLA deadline based on the ticket's priority. This is a rule-based arithmetic calculation, not AI.

First, we'll define some sample SLA policies, similar to how they would be stored in your `sla_policies` table in Supabase. These policies will map a `priority` to a `response_minutes` and `resolution_minutes`.

```python
from datetime import datetime, timedelta

# Sample SLA Policies (based on sla_policies table schema)
sla_policies = {
    "Critical": {"response_minutes": 60, "resolution_minutes": 240},  # 1 hour response, 4 hours resolution
    "High": {"response_minutes": 240, "resolution_minutes": 480},    # 4 hours response, 8 hours resolution
    "Medium": {"response_minutes": 1440, "resolution_minutes": 2880},  # 24 hours response, 48 hours resolution
    "Low": {"response_minutes": 4320, "resolution_minutes": 8640}     # 3 days response, 6 days resolution
}

def calculate_sla_deadlines(priority: str, created_at: datetime) -> dict:
    """Calculates response and resolution due dates based on ticket priority and creation time."""
    policy = sla_policies.get(priority)
    if not policy:
        # Default to a generic policy if priority is not found, or raise an error
        print(f"Warning: SLA policy not found for priority '{priority}'. Using default 'Medium'.")
        policy = sla_policies.get("Medium") # Fallback to a default policy

    response_due_at = created_at + timedelta(minutes=policy["response_minutes"])
    resolution_due_at = created_at + timedelta(minutes=policy["resolution_minutes"])

    return {
        "response_due_at": response_due_at,
        "resolution_due_at": resolution_due_at
    }

print("SLA policies and calculation function defined.")
```

Now, let's use the `calculate_sla_deadlines` function to determine the SLA deadlines for the first ticket loaded from your `tickets.json` file. We'll use its `priority` and `date_received`.

```python
if 'df_tickets' in locals() and not df_tickets.empty:
    # Get priority and date_received from the first ticket in the DataFrame
    ticket_priority = df_tickets['priority'].iloc[0]
    # Convert date_received string to datetime object
    ticket_created_at_str = df_tickets['date_received'].iloc[0]
    ticket_created_at = datetime.strptime(ticket_created_at_str, '%Y-%m-%d %H:%M:%S')

    # Calculate SLA deadlines
    deadlines = calculate_sla_deadlines(ticket_priority, ticket_created_at)

    print(f"Ticket Priority: {ticket_priority}")
    print(f"Ticket Created At: {ticket_created_at}")
    print(f"Response Due At: {deadlines['response_due_at']}")
    print(f"Resolution Due At: {deadlines['resolution_due_at']}")
else:
    print("No tickets DataFrame available or it is empty. Cannot calculate SLA deadlines.")
```

This demonstrates how SLA deadlines would be calculated. In your full system, these calculated `response_due_at` and `resolution_due_at` values would be stored in the `sla_state` table in Supabase, linked to the `ticket_id`.

### Implementing Auto-Reply Check

According to Section 8.6 of your project documentation, the 'Auto-Reply Check' uses `sentence-transformers` to embed the redacted ticket text and compare it against a reference bank of past ticket-reply pairs. If a similarity score exceeds a confidence threshold, an auto-reply is suggested.

```python
# Install sentence-transformers
!pip install sentence-transformers
```

Next, we'll load a pre-trained `sentence-transformers` model and define a simple 'knowledge base' of common questions and their corresponding replies. In a real system, this knowledge base would be dynamic and stored in your Supabase database.

```python
from sentence_transformers import SentenceTransformer, util

# Load a pre-trained sentence transformer model
# 'all-MiniLM-L6-v2' is a good balance of speed and performance
sentence_model = SentenceTransformer('all-MiniLM-L6-v2')

# Simulate a knowledge base (past ticket-reply pairs)
# In a real system, these would be loaded from a database like Supabase
knowledge_base = [
    {
        "question": "My internet is not working. How can I fix it?",
        "reply": "Please try restarting your router and modem. If the issue persists, contact technical support."
    },
    {
        "question": "How do I change my billing address?",
        "reply": "You can update your billing address in your account settings under the 'Billing Information' section."
    },
    {
        "question": "I have a question about my latest bill.",
        "reply": "For billing inquiries, please provide your account number, and our finance department will assist you."
    },
    {
        "question": "I want to report an issue with the mobile app.",
        "reply": "Please describe the issue with the mobile app in detail, including your device type and app version. Our IT department will investigate."
    },
    {
        "question": "What are your operating hours?",
        "reply": "Our customer support operates 24/7. However, specialized departments like Finance or IT may have specific hours."
    }
]

# Pre-compute embeddings for the knowledge base questions
kb_questions = [item['question'] for item in knowledge_base]
kb_embeddings = sentence_model.encode(kb_questions, convert_to_tensor=True)

print("Sentence Transformer model loaded and knowledge base prepared.")
```

Now, let's create a function to compare a new ticket's text against our knowledge base and suggest an auto-reply if the similarity exceeds a set threshold.

```python
def suggest_auto_reply(ticket_text: str, threshold: float = 0.7) -> dict:
    """Suggests an auto-reply for a ticket based on similarity to knowledge base."""
    ticket_embedding = sentence_model.encode(ticket_text, convert_to_tensor=True)

    # Compute cosine similarity between ticket and all KB questions
    cosine_scores = util.cos_sim(ticket_embedding, kb_embeddings)[0]

    # Find the best match
    best_match_idx = torch.argmax(cosine_scores).item()
    max_similarity = cosine_scores[best_match_idx].item()

    if max_similarity >= threshold:
        return {
            "suggested_reply": knowledge_base[best_match_idx]['reply'],
            "similarity_score": round(max_similarity, 3),
            "confidence_threshold_met": True
        }
    else:
        return {
            "suggested_reply": None,
            "similarity_score": round(max_similarity, 3),
            "confidence_threshold_met": False
        }

print("Auto-reply suggestion function defined.")
```

Finally, let's apply the `suggest_auto_reply` function to our sample ticket's redacted body (`sample_ticket_from_file`) and see if an auto-reply is suggested.

```python
if 'sample_ticket_from_file' in locals():
    auto_reply_result = suggest_auto_reply(sample_ticket_from_file, threshold=0.6)

    print(f"Original Ticket Body:\n'{sample_ticket_from_file}'\n")
    print(f"Auto-Reply Check Result: {auto_reply_result}")

    if auto_reply_result['confidence_threshold_met']:
        print(f"\nSuggested Auto-Reply: {auto_reply_result['suggested_reply']}")
    else:
        print("\nNo auto-reply suggested for this ticket as similarity was below threshold.")
else:
    print("Sample ticket text not available. Please ensure previous steps (loading data) were executed.")
```

This completes the implementation of the `Auto-Reply Check`. The `suggest_auto_reply` function would be called as part of the ticket processing pipeline in your FastAPI backend, and the suggested reply (if any) would be stored in Supabase and potentially shown to the customer.

### Orchestrating the Full Ticket Processing Pipeline

Now, let's combine all the components we've built to simulate the end-to-end processing of a single customer support ticket, following the `Architecture Flow of Ticket Lifecycle` outlined in your project document (Section 8). This will bring together PII redaction (simulated), classification, routing, SLA assignment, and auto-reply suggestion.

```python
from datetime import datetime

# Assuming we are processing the first ticket from the loaded DataFrame
if 'df_tickets' in locals() and not df_tickets.empty:
    raw_ticket = df_tickets.iloc[0]

    # --- 1. Ticket Submitted (Simulated from DataFrame) ---
    ticket_id = raw_ticket['ticket_id']
    subject = raw_ticket['subject']
    body = raw_ticket['body'] # Original body
    created_at_str = raw_ticket['date_received']
    # Convert string to datetime object for SLA calculation
    created_at = datetime.strptime(created_at_str, '%Y-%m-%d %H:%M:%S')

    print(f"--- Processing Ticket ID: {ticket_id} ---")
    print(f"Subject: {subject}")
    print(f"Original Body: {body[:150]}...\n")

    # --- 2. PII Redaction (Simulated: using 'body' as already-redacted text) ---
    # In a real system, a dedicated PII redaction function would be called here
    # For this simulation, we'll assume the 'body' column is the redacted text.
    body_redacted = body # Assuming 'body' from your data is already redacted or we redact it here
    print("2. PII Redaction: (Skipped/Simulated - using 'body' as redacted text)\n")

    # --- 3. Classification (DistilBERT) ---
    # Ensure tokenizer and model are loaded
    if 'tokenizer' not in locals() or 'model' not in locals():
        print("Tokenizer or model not found. Re-loading them for full pipeline.")
        tokenizer, model = load_distilbert_model_and_tokenizer()

    classification_results = classify_ticket(body_redacted, tokenizer, model)
    predicted_category = classification_results['category']
    predicted_sentiment = classification_results['sentiment']
    classification_confidence = classification_results['confidence']
    print(f"3. Classification: Category='{predicted_category}', Sentiment='{predicted_sentiment}', Confidence={classification_confidence}\n")

    # --- 4. Routing Engine ---
    assigned_department = route_ticket_to_department(predicted_category)
    print(f"4. Routing Engine: Assigned to '{assigned_department}' department\n")

    # --- 5. SLA Deadline Assigned ---
    # Note: The 'priority' from the classification model might not align with the dataset's 'priority'.
    # Here we use the *predicted* category to *infer* a priority for SLA, or use a default.
    # For simplicity, we'll try to map predicted category to a generic priority, or use the raw_ticket['priority']
    # For more accuracy, a separate model for priority prediction would be used or it's part of the classification.
    # For this demo, let's use the actual priority from the loaded data for SLA calculation
    ticket_priority_for_sla = raw_ticket['priority']
    sla_deadlines = calculate_sla_deadlines(ticket_priority_for_sla, created_at)
    print(f"5. SLA Deadlines: Priority='{ticket_priority_for_sla}'\n   Response Due: {sla_deadlines['response_due_at']}\n   Resolution Due: {sla_deadlines['resolution_due_at']}\n")

    # --- 6. Auto-Reply Check ---
    auto_reply_suggestion = suggest_auto_reply(body_redacted, threshold=0.6)
    print(f"6. Auto-Reply Check: {auto_reply_suggestion['similarity_score']} (Threshold Met: {auto_reply_suggestion['confidence_threshold_met']})")
    if auto_reply_suggestion['confidence_threshold_met']:
        print(f"   Suggested Reply: {auto_reply_suggestion['suggested_reply']}\n")
    else:
        print("   No auto-reply suggested.\n")

    # --- 7. Ticket Stored (Simulated Aggregation) ---
    # This dictionary represents the data that would be stored in Supabase
    processed_ticket_data = {
        "ticket_id": ticket_id,
        "subject": subject,
        "body_redacted": body_redacted,
        "created_at": created_at,
        "predicted_category": predicted_category,
        "predicted_sentiment": predicted_sentiment,
        "classification_confidence": classification_confidence,
        "assigned_department": assigned_department,
        "priority_for_sla": ticket_priority_for_sla,
        "response_due_at": sla_deadlines['response_due_at'],
        "resolution_due_at": sla_deadlines['resolution_due_at'],
        "suggested_auto_reply": auto_reply_suggestion['suggested_reply'] if auto_reply_suggestion['confidence_threshold_met'] else None,
        "auto_reply_confidence": auto_reply_suggestion['similarity_score']
    }

    print("--- Simulated Final Processed Ticket Data (for Supabase storage) ---")
    for key, value in processed_ticket_data.items():
        print(f"  {key}: {value}")
    print("\n--- End of Ticket Processing Simulation ---")

else:
    print("DataFrame 'df_tickets' is not available or empty. Cannot run full ticket processing pipeline.")
```

### Model Evaluation: Confusion Matrix and Accuracy

To evaluate the `classify_ticket` function, we'll use the tickets from the loaded `df_tickets` DataFrame. Since our demonstration DistilBERT model predicts specific categories ('Technical Support', 'Billing Inquiry', 'General Feedback'), we need to map the existing `df_tickets` categories to these for a meaningful comparison.

Here's the mapping strategy I'll apply for evaluation purposes:
*   **'Billing', 'Loan', 'Mortgage', 'Credit card'** from `df_tickets` will be mapped to **'Billing Inquiry'**.
*   Any other categories from `df_tickets` will be mapped to **'General Feedback'** for this demonstration, representing less specific inquiries.
*   We will then classify each ticket body using our DistilBERT model and compare the predicted categories against these mapped actual categories.

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
import numpy as np

# Ensure tokenizer and model are loaded
if 'tokenizer' not in globals() or 'model' not in globals():
    print("Tokenizer or model not found. Loading them...")
    tokenizer, model = load_distilbert_model_and_tokenizer()

# Prepare actual categories by mapping df_tickets categories
def map_category_for_evaluation(original_category):
    if original_category in ['Billing', 'Loan', 'Mortgage', 'Credit card']:
        return 'Billing Inquiry'
    elif original_category in ['Technical Support']:
        return 'Technical Support'
    else:
        return 'General Feedback' # Default for others like 'Other', 'General Feedback'

if 'df_tickets' in globals() and not df_tickets.empty:
    df_eval = df_tickets.copy()
    df_eval['mapped_category'] = df_eval['category'].apply(map_category_for_evaluation)

    # Classify all tickets
    predicted_categories = []
    for index, row in df_eval.iterrows():
        classification = classify_ticket(row['body'], tokenizer, model)
        predicted_categories.append(classification['category'])

    df_eval['predicted_category'] = predicted_categories

    # Filter to only categories that the model can reasonably predict
    # (i.e., those in the category_labels dict from classify_ticket function)
    model_output_categories = ["Technical Support", "Billing Inquiry", "General Feedback"]
    df_filtered = df_eval[df_eval['mapped_category'].isin(model_output_categories)]

    actual_labels = df_filtered['mapped_category'].tolist()
    predicted_labels = df_filtered['predicted_category'].tolist()

    # Calculate Accuracy
    accuracy = accuracy_score(actual_labels, predicted_labels)
    print(f"\nAccuracy: {accuracy:.2f}")

    # Calculate Confusion Matrix
    # Define all possible labels in the order for the confusion matrix
    labels = sorted(list(set(actual_labels + predicted_labels)))
    cm = confusion_matrix(actual_labels, predicted_labels, labels=labels)

    print("\nConfusion Matrix:")
    print(labels) # Print labels to understand matrix rows/columns
    print(cm)

    # Optional: Classification Report for more detailed metrics
    print("\nClassification Report:")
    print(classification_report(actual_labels, predicted_labels, target_names=labels))

else:
    print("DataFrame 'df_tickets' is not available or empty. Cannot perform evaluation.")
```

The confusion matrix above shows how well our (demonstration) DistilBERT model is classifying the tickets based on the defined category mapping. The rows represent the actual mapped categories, and the columns represent the predicted categories.

*   **Accuracy:** The overall accuracy score indicates the proportion of correctly classified tickets.
*   **Confusion Matrix:** Each cell `(i, j)` indicates the number of tickets that actually belonged to category `i` but were predicted as category `j`. Ideally, the diagonal elements should be high, and off-diagonal elements should be low.
*   **Classification Report:** Provides more detailed metrics like precision, recall, and F1-score for each category, which are crucial for understanding the model's performance beyond just overall accuracy.

Keep in mind that this evaluation is based on a general pre-trained DistilBERT model and a simplified category mapping. Fine-tuning the model with a custom dataset tailored to your specific ticket categories and their nuances would significantly improve these metrics in a real-world scenario.
